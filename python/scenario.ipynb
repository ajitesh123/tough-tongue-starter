```python
import requests
import json
```

```python
# API_KEY = "wFBO6IHJZFQLqaZrc1KmxHg3w1se-xIyVLqtzx0zBz8"
API_KEY = "LiMM9Wj2hcz9WHmMfHE7Qv5Kg8U0vrxJUZ1QM8_841o"
API_URL = "https://api.toughtongueai.com/api/public/scenarios"
```

```python
def create_scenario(title, description, role, context, tags=None, media_url=None):
    headers = {
        "Authorization": f"Bearer {API_KEY}",
        "Content-Type": "application/json"
    }

    payload = {
        "name": title,
        "description": description,
        "ai_instructions": role,
        "context": context
    }

    if tags:
        payload["tags"] = tags
    if media_url:
        payload["media_url"] = media_url

    response = requests.post(API_URL, headers=headers, data=json.dumps(payload))

    if response.status_code == 200:
        print("✅ Scenario created successfully!")
        return response.json()
    else:
        print(f"❌ Failed to create scenario. Status Code: {response.status_code}")
        print(f"Response: {response.text}")
        return None

```
